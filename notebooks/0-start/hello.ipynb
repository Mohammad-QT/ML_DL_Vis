## Load a table

The data for the course lives in this project's GitHub repository. `pandas` reads it straight from the web, so there is nothing to upload. `head()` shows the first five rows.

```python
import numpy as np
import pandas as pd

# The data lives in this project's GitHub repository, so there is nothing to upload.
DATA_URL = "https://raw.githubusercontent.com/Mohammad-QT/ML_DL_Vis/main/datasets/customers.csv"
dataset = pd.read_csv(DATA_URL)
dataset.head()
```

## Make a quick chart

Each dot is a shopper. Blue did not buy and pink bought. Age and salary are the two numbers a model could learn from.

```python
import matplotlib.pyplot as plt

rows = dataset.dropna()
plt.figure(figsize=(6, 4))
plt.scatter(rows['age'], rows['salary'], c=rows['purchased'].map({0: '#2f5bd8', 1: '#d6336c'}))
plt.xlabel('Age')
plt.ylabel('Salary')
plt.show()
```
